## 3 · Leer la base

La función depende de la extensión del archivo. La tabla completa y los detalles
de cada formato están en la
[referencia de R](https://cjjmdata.github.io/analisis_datos_i/curso/referencia-r.html#abrir-un-archivo-de-datos).

Guarda la base en un objeto llamado **`serie`**. Es el nombre que usan todas las
sesiones del curso para tu base.

Tres situaciones frecuentes:

**El archivo está en una dirección de internet y es `.csv` o `.csv.gz`**

```r
serie <- read_csv("https://direccion/del/archivo.csv.gz")
```

**El archivo está en internet y es Excel, SPSS o un `.zip`**: primero se descarga
a la máquina de Colab.

```r
download.file("https://direccion/del/archivo.zip", "archivo.zip", mode = "wb")
unzip("archivo.zip")
list.files(recursive = TRUE)      # qué archivos salieron del zip, también dentro de carpetas
serie <- read_csv("nombre_del_archivo.csv")
```

**Descargaste el archivo a tu computadora**: súbelo con el icono de carpeta de la
barra izquierda, y léelo por su nombre.

```r
serie <- read_excel("archivo.xlsx")
```
