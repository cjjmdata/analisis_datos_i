```r
URL <- "https://docs.google.com/spreadsheets/d/e/2PACX-1vTTKC57eWZVIQ9lwhoR5nVqYM4kgi5zA9yifa-YStdfdJwNe7ATs0p-TUCUwvjfcWmHmvsZDEK8VX4I/pub?gid=1326008067&single=true&output=csv"grupo <- read_csv(URL) |>  select(    momento  = Timestamp,    genero   = `Género`,    carrera  = Carrera,    edad     = `Edad, en años cumplidos`,    estatura = `Estatura, en metros`,    traslado = `Tiempo de traslado a la universidad, en minutos`,    calzado  = `Número de calzado`  )
```

## 4 · La primera gráficaNo hace falta entender todavía cómo se construye. Cópiala, córrela y léela: lagramática de `ggplot` es la sesión 6.Las barras van horizontales porque los nombres de las carreras son largos y en eleje de abajo se encimarían.

```r
ggplot(grupo, aes(y = carrera)) +  geom_bar() +  labs(title = "Cómo está compuesto el grupo",       x = "Estudiantes", y = NULL)
```

## 5 · Una comparación entre gruposPromedio de estatura y de calzado, por género.

```r
grupo |>  group_by(genero) |>  summarise(    n        = n(),    estatura = round(mean(estatura, na.rm = TRUE), 2),    calzado  = round(mean(calzado,  na.rm = TRUE), 1)  )
```
