## 2 · Leer los datos del grupo

`read_csv()` lee la hoja donde cayeron sus respuestas.

Lo que sigue a `select()` merece atención, porque hace tres cosas a la vez:

- **renombra** las columnas, que llegan con el título completo de cada pregunta
- **reordena** al orden del cuestionario
- **descarta** la columna de una pregunta que se borró del formulario

Los nombres y el orden de las columnas los decide el análisis, no la fuente.

```r
URL <- "https://docs.google.com/spreadsheets/d/e/2PACX-1vTTKC57eWZVIQ9lwhoR5nVqYM4kgi5zA9yifa-YStdfdJwNe7ATs0p-TUCUwvjfcWmHmvsZDEK8VX4I/pub?gid=1326008067&single=true&output=csv"

grupo <- read_csv(URL) |>
  select(
    momento  = Timestamp,
    genero   = `Género`,
    carrera  = Carrera,
    edad     = `Edad, en años cumplidos`,
    estatura = `Estatura, en metros`,
    traslado = `Tiempo de traslado a la universidad, en minutos`,
    calzado  = `Número de calzado`
  )
```

## 4 · La primera gráfica

No hace falta entender todavía cómo se construye. Cópiala, córrela y léela: la
gramática de `ggplot` es la sesión 6.

Las barras van horizontales porque los nombres de las carreras son largos y en el
eje de abajo se encimarían.

```r
ggplot(grupo, aes(y = carrera)) +
  geom_bar() +
  labs(title = "Cómo está compuesto el grupo",
       x = "Estudiantes", y = NULL)
```

## 5 · Una comparación entre grupos

Promedio de estatura y de calzado, por género.

```r
grupo |>
  group_by(genero) |>
  summarise(
    n        = n(),
    estatura = round(mean(estatura, na.rm = TRUE), 2),
    calzado  = round(mean(calzado,  na.rm = TRUE), 1)
  )
```
